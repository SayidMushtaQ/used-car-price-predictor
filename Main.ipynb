# 🧠 Natural Language Processing (NLP)

## 📌 What is NLP?

**NLP (Natural Language Processing)** is a branch of **Artificial Intelligence (AI)** that helps computers understand, process, analyze, and generate **human language**.

## 🎯 Main Goals of NLP

- Understand human language
- Process and analyze text
- Extract useful information
- Understand sentiment and emotions
- Translate between languages
- Generate human-like text

## 🔧 Applications of NLP

- 🤖 Chatbots
- 🌐 Machine Translation
- 😊 Sentiment Analysis
- 📧 Spam Detection
- 🔍 Search Engines
- 📝 Text Summarization
- 🎙️ Speech Recognition
- ❓ Question Answering

## 🔄 Basic NLP Pipeline

Raw Text  
↓  
Text Cleaning  
↓  
Tokenization  
↓  
Stopword Removal  
↓  
Stemming / Lemmatization  
↓  
Feature Extraction  
↓  
Machine Learning / Deep Learning  
↓  
Prediction / Output

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv('train.txt',sep=';',header=None,names=['text','emotion'])
```

```python
unique_emotion = df['emotion'].unique()

emotion_numbers = {}
i = 0 

for emo in unique_emotion:
    emotion_numbers[emo] = i 
    i+=1 

df['emotion'] = df['emotion'].map(emotion_numbers)
```

```python
df['text']  = df['text'].apply(lambda x: x.lower())
```

```python
df.sample(5)
```

```python
import string 

def remove_punc(txt): 
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

```python
df.sample(5)
```

```python
def remove_numbers(txt): 
    new = ''
    for i in txt: 
        if not i.isdigit(): 
            new = new + i 
    return new 
```

```python
df['text']  = df['text'].apply(remove_numbers)
```

```python
df.head()
```

```python
def remove_emojis(txt): 
    new = ''
    for i in txt: 
        if  i.isascii(): 
            new = new + i 
    return new 
```

```python
df['text']  = df['text'].apply(remove_emojis)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
len(stop_words)
```

```python
def remove_stop_words(txt):
    words = word_tokenize(txt)
    cleaned = []
    for i in words:
        if i not in stop_words: 
            cleaned.append(i)
    return ' '.join(cleaned)
```
